```python
import os
from collections import Counter

label_path = "/content/dataset/train/labels"

class_counts = Counter()

for file in os.listdir(label_path):
    if file.endswith(".txt"):
        with open(os.path.join(label_path, file), "r") as f:
            for line in f:
                class_id = int(line.split()[0])
                class_counts[class_id] += 1

# Replace these names if your dataset has different class names
class_names = {
    0: "Garbage",
    1: "Pothole",
    2: "Street Light",
    3: "Transformer",
    4: "Water Logging"
}

print("Objects per class:\n")

for i in sorted(class_names.keys()):
    print(f"{class_names[i]}: {class_counts[i]}")
```

```python
yaml_content = """

train: /content/civic_dataset/train/images
val: /content/civic_dataset/valid/images
test: /content/civic_dataset/test/images


nc: 5

names:
  0: Garbage
  1: Pothole
  2: street_light
  3: transformer
  4: water_logging

"""


with open("/content/civic_road_issues.v1-civic_issues5.yolov8.zip","w") as f:
    f.write(yaml_content)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")   # Use yolov8s instead of yolov8n
```

```python
model.train(
    data="/content/dataset/data.yaml",
    epochs=60,
    imgsz=640,
    batch=16,
    optimizer="SGD",
    patience=15,
    cache=True,
    workers=2,
    device=0
)
```

```python
metrics = model.val()

print(metrics)
```

```python
import os

for root, dirs, files in os.walk("/content/runs"):
    for file in files:
        if file.endswith(".pt"):
            print(os.path.join(root,file))
```

```python
best_model = YOLO(
    "/content/runs/detect/train/weights/best.pt"
)
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
results = best_model(
    "pothole.jpg",
    save=True
)


civira_categories = [
    "Garbage",
    "Pothole",
    "street_light",
    "water_logging",
    "transformer"
]


for result in results:


    if len(result.boxes) == 0:

        print("❌ Irrelevant Image")


    else:

        for box in result.boxes:

            class_id = int(box.cls[0])

            confidence = float(box.conf[0])


            category = best_model.names[class_id]


            if category in civira_categories:

                print("✅ Relevant Civic Issue")

                print(
                    "Category:",
                    category
                )

                print(
                    "Confidence:",
                    round(confidence*100,2),
                    "%"
                )

            else:

                print("❌ Irrelevant")
```
